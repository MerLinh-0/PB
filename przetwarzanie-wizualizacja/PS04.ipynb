```python
# 2. Stwórz 4 kolumny z wartościami na podstawie kolumny daty:  dzień, miesiąc i rok z tej daty, jak również dzień tygodnia.

df["day"] = df["Date"].dt.day
df["month"] = df["Date"].dt.month
df["year"] = df["Date"].dt.year
df["day_of_week"] = df["Date"].dt.day_of_week

df
```

```python
# 4. Wydobądź wszystkie pomidory (“Tomato”) z listy wszystkich produktów/dóbr zbioru danych kalimati (a jest tam kilka gatunków). (Warunek z wykorzystaniem odpowiedniej funkcji dla danych tekstowych)

tomatos = df.loc[df["Commodity"].str.startswith("Tomato"), "Commodity"].unique()
tomatos
```

```python
# 5. Odfiltruj informacje dla jedngo z dóbr nie będącego kiwi opisanego w zbiorze danych. W wynikowej ramce danych dodaj kolumnę ze zmianą ceny danego dnia od poprzedniego notowania.

df_pumpkin = df.loc[df["Commodity"] == "Pumpkin"]
df_pumpkin["average_diff"] = df.loc[df["Commodity"] == "Pumpkin", "Average"].diff(1)
#pd.set_option("display.max_rows", None)
df_pumpkin
```

```python
# 6. Na podstawie wyjściowego zbioru danych utwórz ramkę danych dla wszystkich gatunków pomidorów zawierającą kolumny (proszę zaproponować krótkie nazwy): 
#    - nazwa produktu, 
#    - data najniższego notowania (najniższej ceny), 
#    - najniższa notowana cena, 
#    - data notowania najwyższej ceny, 
#    - najwyższa cena, 
#    - pierwszy dzień notowania (data), 
#    - ostatni dzień notowania (data),
#    - średnia ceny ze wszystkich notowanych dni, 
#    - liczba dni notowania danego produktu i
#    - średnia cen ze wszystkich notowań w piątki.

df_tomatos = df.loc[df["Commodity"].str.startswith("Tomato")]

min_idx = df_tomatos.groupby("Commodity")["Minimum"].idxmin()
min_date = df_tomatos.loc[min_idx, ["Commodity", "Date", "Minimum"]]

max_idx = df_tomatos.groupby("Commodity")["Maximum"].idxmax()
max_date = df_tomatos.loc[max_idx, ["Commodity", "Date", "Maximum"]]

first_day_idx = df_tomatos.groupby("Commodity")["Date"].idxmin()
first_day = df_tomatos.loc[first_day_idx, ["Commodity", "Date"]]

last_day_idx = df_tomatos.groupby("Commodity")["Date"].idxmax()
last_day = df_tomatos.loc[last_day_idx, ["Commodity", "Date"]]

avg_alltime = df_tomatos.groupby("Commodity")["Average"].agg("mean")

df_tomatos_friday = df_tomatos.loc[df_tomatos["day_of_week"] == 4]
avg_friday = df_tomatos_friday.groupby("Commodity")["Average"].agg("mean")

df_merge = min_date.merge(max_date, on="Commodity", suffixes=("_min", "_max")).merge(first_day, on="Commodity").merge(last_day, on="Commodity").merge(avg_alltime, on="Commodity")
df_merge["num_of_days"] = df_merge["Date_y"] - df_merge["Date_x"]
df_merge = df_merge.merge(avg_friday, on="Commodity", suffixes=("_alltime", "_friday"))
df_merge
```
