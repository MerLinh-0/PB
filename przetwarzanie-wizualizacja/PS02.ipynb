Proste zestawienia tworzymy za pomocą metody `value_counts()`. Pozwala na poznanie rozkładu danych dyskretnych. Można podejrzeć czy nie ma pojedyńczych wartości w danych, które są nieprawidłowo wpisane. Domyślnie pomija wartości brakujące, jeśli występują.

```python
df["parental level of education"].value_counts()
```

```python
df["parental level of education"].value_counts(dropna=False) # bez pomijania wartości brakujących
```

W przypadku danych ciągłych (liczbowych) nie ma za wiele sensu, ponieważ nic ciekawego z tego nie możemy odczytać.

```python
df["parental income"].value_counts()
```

Aby podglądać proste zależności między zmiennymi dyskretnymi oraz reprezentatywność podgrup możemy wykorzystać `pd.crosstab()`

```python
pd.crosstab(df["parental level of education"], df["years to graduate"])
```

`pd.crosstab()` przyjmuje jako dwa pierwsze parametry serie lub listy serii danych, pierwsza reprezentuje zmienne podsumowywane w wierszach a druga zmienne podsumowywane w kolumnach. Obydwa parametry muszą być tej samej długości lub tej samej liczby wierszy, ponieważ funkcja ta zestawia te dane ze sobą i zlicza poszczególne kombinacje wartości.

aby to zaprezentować dodajmy zmienną z zaokrągloną średnią do wartości 0.5.

```python
df["rounded GPA"] = (df["high school gpa"]*2).round()/2
pd.crosstab([df["parental level of education"],df["rounded GPA"]], df["years to graduate"])
```

```python
pd.crosstab(df["years to graduate"], [df["parental level of education"],df["rounded GPA"]])
```

W powyższych przykładach możliwe było jedynie zliczanie wystąpień poszczególnych wartości lub kombinacji wartości. Często chcemy jednak poznać podsumowanie poszczególnych grup ze względu na zmienne (kolumny) zawarte w ramce danych.

Prosty przykład ze średnią oceną w szkole średniej dla studentów o rodzicach z różnym wykształceniem

```python
df.groupby("parental level of education")["high school gpa"].mean()
```

Średnie plasują się dość intuicyjnie. Dla tych co mają rodziców z lepszym wykształceniem średnia jest nieco wyższa. Jedank proszę pamiętać, że aby móc wnioskować na temat średnich różnych grup korzystamy z testów statystycznych (odsyłam do przedmiotu **Statystyka matematyczna**).

dugą opcją jest wykorzystanie metody `agg()` zamiast wprost metody funkcji agrefującej

```python
df.groupby("parental level of education")["high school gpa"].agg("mean")
```

zrealizujmy więc to co realizują metody `value_counts()` i `crosstab()`.

```python
print(df.groupby("parental level of education")["high school gpa"].agg("count"))
print("\n")
# lista kolumn po których grupujemy!
print(df.groupby(["parental level of education","years to graduate"])["high school gpa"].agg("count"))
```

Aby zrobić z tego tabelkę podobną do wyniku `crosstab()` należy utworzyć tabelę przestawną, będziemy taką wykonywać na następnych zajęciach.

Zammiast `count()` na kolumnie można wykonać metodę `size()` bez podawania kolumny

```python
print(df.groupby(["parental level of education","years to graduate"]).size())
```

Metoda `agg()` pozwala na policzenie więcej niż jednej statystyki.

Oprócz średniej policz jeszcze medianę i odchylenie standardowe. (więcej funkcji agregujących - vide Wykład)

```python
df.groupby("parental level of education")["high school gpa"].agg(["mean","std","median"])
```

A jak sobie radzą studenci na studiach biorąc pod uwagę wykształcenie ich rodziców i ich średnią ze szkoły średniej? (wykorzystajmy zaokrąglone średnie ze szkoły średniej)

```python
df.groupby(["parental level of education","rounded GPA"])["college gpa"].agg(["mean","std"])
```

Jak widzimy w niektórych przypadkach było za mało danych aby policzyć odchylenie standardowe (potrzeba co najmniej dwóch wierszy w danej kombinacji wartości zmiennych grupujących). W takich przypadkach jest zwracana wartość reprezentująca brak danych a w tabeli widzimy to jako _NaN_ - _not a number_.

Funkcje agregujące dostępne domyślnie z biblioteką często są niewystarczające. Dlatego można wykorzystać funkcje lambda do definiowania swoich złożonych funkcji agregujących.

```python
df.groupby("parental level of education")["high school gpa"].agg(lambda x: max(x)-min(x))
```

Trzecia największa średnia dla studentów o tym samym poziomie wykształcenia rodziców

```python

df.groupby("parental level of education")["high school gpa"].agg(lambda x: x.nlargest(3).iloc[-1])
```
