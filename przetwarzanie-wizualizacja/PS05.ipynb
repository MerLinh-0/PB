Dodatkowo niektóre pliki (np. statystyki przestępczości) posiadają na końcu wierszy puste średniki, co zmusza nas do użycia parametru usecols, aby zignorować śmieciowe, puste kolumny tworzące się na końcu tabeli. ## 2. Czyszczenie pojazdów

Czyszczenie i agregacja danych o zarejestrowanych pojazdach
Dane o pojazdach są rozbite na poszczególne miesiące, a my chcemy mieć jedną wartość dla całego roku. Funkcja clean_pojazdy najpierw naprawia błędy strukturalne plików (dodatkowe separatory na końcu linii tworzące puste kolumny Unnamed i przesuwające indeksy), a następnie usuwa zbędną kolumnę TERYT. Kolejnym problemem w nowszych plikach (od 2021 r.) są liczby zapisywane jako tekst ze spacjami (np. "10 176"). Funkcja usuwa te spacje, konwertuje wartości na typ liczbowy, a na koniec sumuje wszystkie miesiące do jednej wynikowej kolumny dla danego roku (np. rej2018).


```python
def clean_pojazdy(df, rok):
    unnamed = df.columns[df.columns.str.contains("Unnamed:")].to_list()
    empty = lambda col: df[col].isnull().all()
    to_remove = [col for col in unnamed if empty(col)]
    df = df.drop(columns=to_remove)

    df.rename(columns={df.columns[0]: 'wojewodztwo'}, inplace=True)
    df['wojewodztwo'] = df['wojewodztwo'].astype(str).str.lower().str.strip()

    if "TERYT" in df.columns:
        df = df.drop(columns=["TERYT"])

    wojewodztwa_lista = [
        'dolnośląskie', 'kujawsko-pomorskie', 'lubelskie', 'lubuskie', 
        'łódzkie', 'małopolskie', 'mazowieckie', 'opolskie', 
        'podkarpackie', 'podlaskie', 'pomorskie', 'śląskie', 
        'świętokrzyskie', 'warmińsko-mazurskie', 'wielkopolskie', 'zachodniopomorskie'
    ]

    df = df[df['wojewodztwo'].isin(wojewodztwa_lista)].copy()
    
    kolumny_miesiace = df.columns.drop('wojewodztwo')
    for kol in kolumny_miesiace:
        df[kol] = df[kol].astype(str).str.replace(' ', '', regex=False).astype(float)

    nazwa_wynikowa = f'rej{rok}'
    df_wynik = df[['wojewodztwo']].copy()
    df_wynik[nazwa_wynikowa] = df[kolumny_miesiace].sum(axis=1)
    return df_wynik

pojazdy_2018 = clean_pojazdy(pojazdy_2018_mies, 2018)
pojazdy_2019 = clean_pojazdy(pojazdy_2019_mies, 2019)
pojazdy_2020 = clean_pojazdy(pojazdy_2020_mies, 2020)
pojazdy_2021 = clean_pojazdy(pojazdy_2021_mies, 2021)
pojazdy_2022 = clean_pojazdy(pojazdy_2022_mies, 2022)
pojazdy_2023 = clean_pojazdy(pojazdy_2023_mies, 2023)

pojazdy_2021_mies
```

## 3. Przekształcanie danych policyjnych

Przekształcanie danych (Kradzieże, Przestępstwa, Nietrzeźwi).
Pliki mają zupełnie inną strukturę niż pozostałe – lata ułożone są w nich pionowo (w kolejnych wierszach), a nie poziomo (w kolumnach). Funkcja pivot ma za zadanie "przestawić" tabelę tak, aby wybrane lata (2018-2023) stały się osobnymi kolumnami. Zanim to jednak nastąpi, funkcja musi oczyścić kolumnę z województwami: usuwa przedrostki "woj.", odcina zbędne dopiski (jak informacje o komendach w woj. mazowieckim) i sprowadza wszystko do małych liter, aby nazwy pasowały do znormalizowanej listy.

```python
wojewodztwa_lista = [
        'dolnośląskie', 'kujawsko-pomorskie', 'lubelskie', 'lubuskie', 
        'łódzkie', 'małopolskie', 'mazowieckie', 'opolskie', 
        'podkarpackie', 'podlaskie', 'pomorskie', 'śląskie', 
        'świętokrzyskie', 'warmińsko-mazurskie', 'wielkopolskie', 'zachodniopomorskie'
    ]

def pivot(df, column, prefix):
    df['wojewodztwo'] = df['wojewodztwo'].astype(str).str.replace('woj.', '', regex=False).str.split('(').str[0].str.strip().str.lower()
    df = df[df['wojewodztwo'].isin(wojewodztwa_lista)]

    df[column] = df[column].astype(str).str.replace(' ', '', regex=False)
    df[column] = pd.to_numeric(df[column], errors='coerce')

    df_lata = df[df['rok'].isin([2018, 2019, 2020, 2021, 2022, 2023])]

    df_pivot = df_lata.pivot(
        index='wojewodztwo', 
        columns='rok', 
        values=column
    ).reset_index()

    df_pivot.columns = ['wojewodztwo', f'{prefix}2018', f'{prefix}2019', f'{prefix}2020', f'{prefix}2021', f'{prefix}2022', f'{prefix}2023']
    return df_pivot

kradziez_pivot = pivot(kradziez, "kradzieze_stwierdzone", "krad")
nietrzezwi_pivot = pivot(nietrzezwi, "nietrzezwi_stwierdzeni", "nietrze")
kryminal_pivot = pivot(przestepstwa, "kryminal_stwierdzone", "kryminal")
kryminal_pivot
```

## 4. Czyszczenie zdarzeń drogowych

Ujednolicanie nazw w zdarzeniach drogowych.
W plikach dotyczących zdarzeń drogowych (wypadki, zabici, ranni) wartości liczbowe wczytują się poprawnie, jednak nazwy województw są zapisane wielkimi literami i poprzedzone przedrostkiem (np. "WOJ. DOLNOŚLĄSKIE"). Funkcja clean_zdarzenia usuwa ten przedrostek, zamienia litery na małe i filtruje tabele tak, aby pozbyć się podsumowań ogólnopolskich i pustych wierszy z końca plików.

```python
def clean_zdarzenia(df, rok):
    df.columns = ['wojewodztwo', f'wypadki{rok}', f'zabici{rok}', f'ranni{rok}', f'kolizje{rok}']
    df['wojewodztwo'] = df['wojewodztwo'].astype(str).str.replace('WOJ. ', '', regex=False).str.strip().str.lower()
    df = df[df['wojewodztwo'].isin(wojewodztwa_lista)].copy()
    return df

zdarzenia_2018 = clean_zdarzenia(zdarzenia_2018, 2018)
zdarzenia_2019 = clean_zdarzenia(zdarzenia_2019, 2019)
zdarzenia_2020 = clean_zdarzenia(zdarzenia_2020, 2020)
zdarzenia_2021 = clean_zdarzenia(zdarzenia_2021, 2021)
zdarzenia_2022 = clean_zdarzenia(zdarzenia_2022, 2022)
zdarzenia_2023 = clean_zdarzenia(zdarzenia_2023, 2023)
```

## 5. Łączenie wszystkiego

Ostatnim etapem jest połączenie wszystkich wyczyszczonych i zagregowanych ramek danych w jedną, szeroką tabelę.

```python
# Bazową tabelą (lewą) staje się pierwsza z naszych ramek, do której doklejamy kolejne
df_final = (
    pojazdy_2018
    .merge(pojazdy_2019, on='wojewodztwo', how='outer')
    .merge(pojazdy_2020, on='wojewodztwo', how='outer')
    .merge(pojazdy_2021, on='wojewodztwo', how='outer')
    .merge(pojazdy_2022, on='wojewodztwo', how='outer')
    .merge(pojazdy_2023, on='wojewodztwo', how='outer')
    
    .merge(zdarzenia_2018, on='wojewodztwo', how='outer')
    .merge(zdarzenia_2019, on='wojewodztwo', how='outer')
    .merge(zdarzenia_2020, on='wojewodztwo', how='outer')
    .merge(zdarzenia_2021, on='wojewodztwo', how='outer')
    .merge(zdarzenia_2022, on='wojewodztwo', how='outer')
    .merge(zdarzenia_2023, on='wojewodztwo', how='outer')
    
    .merge(kradziez_pivot, on='wojewodztwo', how='outer')
    .merge(nietrzezwi_pivot, on='wojewodztwo', how='outer')
    .merge(kryminal_pivot, on='wojewodztwo', how='outer')
)

df_final.info()
df_final
```
