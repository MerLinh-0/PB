```python
import numpy as np
import pandas as pd
import scipy.stats as stats
import matplotlib.pyplot as plt
from scipy.stats import normaltest
from IPython.display import display

plik = "USvideos.csv"
df = pd.read_csv(plik)
display(df.head())
print("Wymiary:", df.shape)
print("Kolumny i typy:")
print(df.dtypes)
```

```python
pd.options.display.float_format = '{:,.4f}'.format
X = df["views"]
Y = df["comments_disabled"]
podsumowanie_X = pd.DataFrame({
    "miara": ["średnia", "mediana", "S²", "S", "Ŝ²", "Ŝ", "min", "max"],
    "X (views)": [
        X.mean(), X.median(), X.var(ddof=0), X.std(ddof=0), X.var(ddof=1), X.std(ddof=1), X.min(), X.max()
    ]
})
display(podsumowanie_X)
```

```python
podsumowanie_Y = pd.DataFrame({
    "miara": ["n", "m (sukcesy)", "n-m", "frakcja", "dopełnienie"],
    "Y (comments_disabled)": [len(Y), np.sum(Y), len(Y)-np.sum(Y), np.sum(Y)/len(Y), 1-(np.sum(Y)/len(Y))]
})
display(podsumowanie_Y)
```
